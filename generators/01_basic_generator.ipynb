# Basic Generator

## Lists vs Generators - Which is more efficient ?

### List of Numbers

Notes:
1. Lists evaluate eagerly
2. All elements in list are stored in memory

```python
list_of_numbers = [0, 1, 2, 3, 4]
```

```python
for i in list_of_numbers:
    print(i)
```

#### an equivalent approach using `range()`

```python
list_of_numbers = [i for i in range(5)]
```

```python
for i in list_of_numbers:
    print(i)
```

### Generator of Numbers

Notes:
1. Generators are evaluated lazily - Only when they are iterated upon
2. Only 1 item (current item) stored in memory

```python
gen = (i for i in range(5))
```

```python
type(gen)
```

```python
for i in gen:
    print(i)
```

## Generator Functions

1. The presence of the `yield` identifes the function as a generator
2. `yield` returns a value upon each iteration

```python
def generator_of_numbers(start: int=0, end: int=1_000_000):
    for i in range(start, end+1):
        yield i
```

```python
for i in generator_of_numbers(1, 5):
    print(i)
```

### another example

```python
def generator_with_multiple_yields():
    yield 1
    yield 'two'
    yield {3: 'three'}
```

```python
for i in generator_with_multiple_yields():
    print(i)
```
