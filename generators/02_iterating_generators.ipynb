# Iterating Generators

## Define a Generator yielding exactly 3 items

```python
def generator_of_numbers():
    for i in range(1, 4):
        yield i
```

### Iterate using a for-loop

```python
gen = generator_of_numbers()

for i in gen:
    print(i)
```

### Iterate using next()

```python
gen = generator_of_numbers()
```

#### Get first item

```python
next(gen)
```

#### Get second item

```python
next(gen)
```

#### Get third item

```python
next(gen)
```

#### Get fourth item (non-existent)

```python
next(gen)
```

1. `StopIteration` - signifies that the generator is exhausted
2. Make sure `StopIteration` is handled using `try-except` when using `next()`
3. `for-loops` automatically handle `StopIteration` and exits the loop

### Suppressing StopIteration in next()

```python
gen = generator_of_numbers()
```
