# Bi-directional Communication in Generators

## Do generators only produce data ? Can they consume data as well ?

### Lets explore an alternate method of iterating generators

So far we have iterated generators using:
1. `next()`
2. `for-loops`

```python
def count_from(n: int):
    while True:
        yield n
        n += 1
```

```python
gen = count_from(100)
```

```python
gen.send(None)
```

```python
gen.send(None)
```

```python
gen.send(None)
```

Notes: 
1. Notice how `send(None)` behaves almost exactly as `next()`
2. Why do we need `send()` if we already have `next()` ?  

### Exploring Bi-directional Communication

On each iteration, generators can both:
1. yield a value
2. accept a value 

```python
def increment():
    n = None
    while True:
        n = yield n
        n += 1
```

```python
gen = increment()
```

```python
gen.send(None)
```

```python
gen.send(5)
```

```python
gen.send(15)
```

### Rolling Average

Notes:
1. This example goes beyond standard data iteration use cases we have seen thus far
2. Showcases versatlity of python generators in acting simulatenously as both a data producer and consumer

```python
def rolling_average(window: int = 3):
    data = []
    ra = None
    while True:
        next_value_from_send = yield ra
        data.append(next_value_from_send)
        data = data[-window:]
        ra = sum(data)/len(data)    
```

```python
rolling_average_generator = rolling_average()
```

Initiallize / Prime the generator
1. Always start with `send(None)`
2. The generator executes until it encounters the first `yield`

```python
rolling_average_generator.send(None)
```

```python
rolling_average_generator.send(1) # Average of [1] = 1
```

```python
rolling_average_generator.send(2) # Average of [1, 2] = 1.5
```

```python
rolling_average_generator.send(3) # Average of [1, 2, 3] = 2
```

```python
rolling_average_generator.send(4) # Average of [2, 3, 4] = 3
```

```python
rolling_average_generator.send(5) # Average of [3, 4, 5] = 4
```
