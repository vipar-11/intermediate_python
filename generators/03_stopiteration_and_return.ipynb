# StopIteration and return statements

## Effect of return statements in a generator

```python
def generator_of_numbers():
    for i in range(1,4):
        yield i
    return 'NO_MORE_NUMBERS'
```

```python
for i in generator_of_numbers():
    print(i)
```

```python
gen = generator_of_numbers()
```

```python
next(gen)
```

```python
next(gen)
```

```python
next(gen)
```

```python
next(gen)
```

Notice `NO_MORE_NUMBERS` within `StopIteration`

### A premature return

```python
def generator_of_numbers2():
    yield 1
    yield 2
    return 'END' #can't get past me !!!
    yield 3 #unreachable
```

```python
gen2 = generator_of_numbers2()
```

```python
next(gen2)
```

```python
next(gen2)
```

```python
next(gen2)
```

```python
next(gen2)
```
