# Generators are stateful

## Stateless Functions vs Stateful Generators

### A stateless function

```python
def next_number(n: int):
    return n+1
```

```python
next_number(5)
```

```python
next_number(10)
```

Notes:
1. Each function call is independent
2. Second call to function has no memory of the first

### Generators are stateful and have memory

```python
def count_from(n: int):
    while True:
        yield n
        n += 1
```

```python
gen = count_from(n=100)
```

#### iterate using next()

```python
next(gen)
```

```python
next(gen)
```

```python
next(gen)
```

Notes:
1. Notice how the value of `n` changes after every iteration step
2. Behaves almost like an instance variable of a class

#### iterate using for-loop

same behavior in `for-loops`

```python
gen2 = count_from(100)

for i in gen2:
    print(i)
    if i > 104:
        break
```
